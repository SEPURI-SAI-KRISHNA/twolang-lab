# `enum`: `auto()`, `Flag` (bitwise combos), `StrEnum`/`IntEnum`

`auto()` assigns values so you don't have to (and don't accidentally duplicate one). `Flag` members combine
with `|` like real bitflags, and support `in` for membership checks -- built for "which of these options are
set," not "which one option is this." `StrEnum` (3.11+) and `IntEnum` members ARE actual `str`/`int`
instances -- they compare equal to plain strings/ints directly, unlike a plain `Enum`.

```python
from enum import Enum, Flag, auto, StrEnum, IntEnum

class Color(Enum):
    RED = auto()
    GREEN = auto()
    BLUE = auto()

print("auto() assigned values:", list(Color))

class Permission(Flag):
    READ = auto()
    WRITE = auto()
    EXECUTE = auto()

rw = Permission.READ | Permission.WRITE
print("combined flags:", rw)
print("READ in rw:", Permission.READ in rw)
print("EXECUTE in rw:", Permission.EXECUTE in rw)
```

```python
class Status(StrEnum):
    ACTIVE = "active"
    INACTIVE = "inactive"

print("StrEnum member IS a str:", isinstance(Status.ACTIVE, str))
print("compares equal to a plain string directly:", Status.ACTIVE == "active")

class Priority(IntEnum):
    LOW = 1
    HIGH = 10

print("IntEnum supports int comparisons directly:", Priority.HIGH > Priority.LOW, Priority.HIGH > 5)
```
